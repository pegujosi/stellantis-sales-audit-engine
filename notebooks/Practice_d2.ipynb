```python
import pandas as pd
import numpy as np

df=pd.read_parquet('../data/cl_data.parquet')
df.head(10)
```

```python
conditions = [df['unidades']==1,(df['unidades']>1)&(df['unidades']<=4),df['unidades']>4]
options = ['individual','medio','alto']
print(type(conditions))
print(type(options))
```

```python
df['cat_unidades'] = np.select(conditions, options, default='No aplica')
df.head(10)
```

```python
total_ventas_brutas = df['venta_bruta'].sum() 
total_unidades_vendidas = df['unidades'].sum()
ticket_promedio = total_ventas_brutas / total_unidades_vendidas

print(f'El total de ventas brutas es: ${total_ventas_brutas:,.1f} pesos')
print(f'El total de unidades vendidas es: {total_unidades_vendidas:,} unidades')
print(f'El ticket promedio es: ${ticket_promedio:,.2f} pesos')
```

```python
top_ventas = df.groupby('id_cliente')['venta_bruta'].sum().sort_values(ascending=False).reset_index().head(10)
print(type(top_ventas))
top_ventas['venta_bruta'] = top_ventas['venta_bruta'].map('${:,.2f} MXN'.format)
top_ventas
```

```python
des_ubicacion = df.groupby('id_ubicacion')['venta_bruta'].agg(desviacion_std_ventas='std',mediana_de_ventas='median',promedio_venta='mean').round().reset_index()
print(type(des_ubicacion['promedio_venta']))
des_ubicacion['coef_variacion_%'] = ((des_ubicacion['desviacion_std_ventas'] / des_ubicacion['promedio_venta']) * 100).round()
des_ubicacion
```

```python
def outliers_IQR(df, serie):
    Q1 = df[serie].quantile(0.25)
    Q3 = df[serie].quantile(0.75) 
    IQR = Q3-Q1

    limite_inf = Q1 - (1.5 * IQR)
    limite_sup = Q3 + (1.5 * IQR)

    outliers = df[(df[serie] < limite_inf) | (df[serie] > limite_sup)]

    print(f'---Analisis IQR para {serie}---')
    print(f' El rango Q1 es {Q1}')
    print(f' El rango Q3 es {Q3}')
    print(f' El IQR es {IQR}')
    print(f' El limite inferior es {limite_inf}')
    print(f' El limite superior es {limite_sup}')
    print(f' Total Outliers detectados: {len(outliers)} ({len(outliers)/len(df):.2%})\n')

    return outliers
```

```python
df_outliers = outliers_IQR(df,'venta_bruta')
#df_outliers.head(10)
```
